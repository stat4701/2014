# Stage B: Pulling apart multiple populations

My domain knowledge of poker tells me that the first step I take with the consolidated data is separating the two populations that exist. I explained during Stage A that there are two types of players, regulars and recreational players. Since this study is about finding the differences between winning and losing __regulars__, now would be a good time to try and get rid of the recreational players.

The two most commonly used statistics in poker are called VP\$IP (Voluntarily Put Money In Pot) and PFR (Preflop Raise). My domain knowledge also tells me these statistics will help us classify regulars and recreational players. 

I think a good approach to this problem is clustering by k-means. Another possibility is treating it as a Gaussian Mixture Model. The aggregate winrates of the different clusters will help us remove certain players.

__Goal__: I'm not interested in recreational players so I want to finish Stage B with a dataframe of regulars.

```python
def login(number):
	db_name = 'EDAV_'+str(number) # the databases are EDAV_1,...,EDAV_4
	connection = psycopg2.connect("dbname=" + db_name + " user=postgres password=postgrespass")
	mark = connection.cursor()
	return mark
```

```python
def rename_headers(table_headers, record): # RENAME COLUMNS
	column_names = 'select column_name from information_schema.columns where table_name = ' + table_headers
	mark.execute(column_names)
	headers = mark.fetchall()

	i=0
	for col in headers:
		col = str(col)[2:-3]
		record.rename(columns={i:col}, inplace=True)
		i += 1

def rename_custom(custom_headers, df): # RENAME COLUMNS

	i=0
	for col in custom_headers:
		df.rename(columns={i:col}, inplace=True)
		i += 1
```

For VP\$IP, the calculation is (cnt_vpip / (cnt_hands - cnt_walks)) * 100:

For PFR, the calculation is (cnt_pfr / cnt_pfr_opp) * 100

Below is a function that will create dataframes for all 4 databases.

```python
def vpip_pfr(number):

    mark = login(number)
    distinct = ''
    column = 'id_player, id_limit, cnt_vpip, cnt_hands, cnt_walks, cnt_pfr, cnt_pfr_opp, amt_expected_bb_won'
    table = 'cash_cache'
    
    custom_headers = ('PLAYER_ID', 'LIMIT', 'cnt_vpip', 'cnt_hands', 'cnt_walks', 'cnt_pfr', 'cnt_pfr_opp', 'amt_expected_bb_won')
    
    statement = 'SELECT ' + column + ' FROM ' + table
    
    mark.execute(statement)
    
    vpip = mark.fetchall()
    vpip = pd.DataFrame(vpip)
    rename_custom(custom_headers, vpip)
    
    ## LIMIT IDs
    statement = 'SELECT id_limit,amt_bb FROM cash_limit'
    
    mark.execute(statement)
    limits = mark.fetchall()
    limits = pd.DataFrame(limits)
    
    custom_headers = ('LIMIT','BB_SIZE')
    rename_custom(custom_headers, limits)
    
    limits.to_csv('limits.csv')
    
    ## PLAYER NAMES
    statement = 'SELECT id_player, player_name FROM player'
    mark.execute(statement)
    screennames = mark.fetchall()
    screennames = pd.DataFrame(screennames)
    
    custom_headers = ('PLAYER_ID', 'NAME')
    rename_custom(custom_headers, screennames)
    
    for i in range(len(screennames.NAME)):
        name = screennames.NAME.ix[i]
        name = unicode(name, errors = 'ignore')
        screennames.NAME.ix[i] = name
    
    screennames.to_csv('screennames.csv')
    
    
    screennames = screennames.set_index('PLAYER_ID')
    limits = limits.set_index('LIMIT')
    
    limits = limits[np.logical_or(limits.BB_SIZE == 1,limits.BB_SIZE == 2)]
    
    vpip = vpip.join(limits, on='LIMIT', how = 'left')
    del vpip['LIMIT']
    
    vpip = vpip[np.logical_or(vpip.BB_SIZE == 1, vpip.BB_SIZE == 2)]
    
    vpip = vpip.join(screennames, on = 'PLAYER_ID', how = 'left')
    vpip = vpip.set_index('NAME')
    del vpip['PLAYER_ID']
    del vpip['BB_SIZE']
    vpip.to_csv('3_VPIP_PFR' + str(number) + '.csv')

for i in range(1,5):
    vpip_pfr(i)
```

Here is the question: for a binomial distribution with mean .3118 (or .1678 in the case of PFR), what is the number of hands we need in order to be within 10% of the true statistic 95% of the time?

This is easy to solve analyitically using the CLT but it's more fun and educational to have python do it for us.

$X_{i,VPIP} \sim Bernoulli(0.3118)$

$X_{i,PFR} \sim Bernoulli(0.1678)$

Note: there is a strong correlation between VP\$IP and PFR, but I don't think looking at them separately will cause any problems in this particular situation.

Another note: This is a bit of a spoiler, but we'll see later on that VP\$IP is a far more important indicator than PFR so we're more concerned about that stat here.

```python
def generate_samples(n):
    from scipy.stats import bernoulli,poisson,norm,expon,binom

    vpip_sample = binom.rvs(n,mean_vpip/100)/float(n)
    pfr_sample = binom.rvs(n,mean_pfr/100)/float(n)

    samples = {}
    for i in range(1000000):
        samples[i] = []
        samples[i].append(binom.rvs(n,mean_vpip/100)/float(n))
        samples[i].append(binom.rvs(n,mean_pfr/100)/float(n))

    df = DataFrame.from_dict(samples, orient = 'index')
    df.columns = 'VPIP','PFR'
    df.index.name = 'SAMPLE'
    return df

df = generate_samples(50)
df.head()
```

There are some nice side benefits to limiting the minimum number of hands:

```python
def compare_random_t(vpip, n, t_params):
    style.use('ggplot')

    randoms = stats.t.rvs(t_params[0], t_params[1], t_params[2], size = len(vpip))
    
    ax = fig.add_subplot(3,3,n)
    x = np.arange(vpip.VPIP.min(),vpip.VPIP.max())
    
    colors = brewer2mpl.get_map('BrBG','diverging',4).mpl_colors
    
    hist(vpip.VPIP, bins = 100, normed = True, cumulative = False, 
         histtype = 'stepfilled', color = colors[3], alpha = 0.75, label = 'Empirical Data')
    hist(randoms, bins = 100, normed = True, cumulative = False, 
         histtype = 'stepfilled', color = colors[0], alpha = 0.35, label = 'Random Data')
    vlines(t_params[1], 0, 0.080, linestyles = '--', label = 'Mean')
    
    if n == 1:
        xlabel('VP\$IP')
        ylabel('Frequency')
        ax.legend(loc = 'best')

    for label in ax.get_xticklines():
        label.set_visible(False)
    for label in ax.get_yticklines():
        label.set_visible(False)
    if n != 1:
        for label in ax.get_xticklabels():
            label.set_visible(False)
        for label in ax.get_yticklabels():
            label.set_visible(False)  

    xlim([0,60])
    ylim([0,0.080])
        
t_params = stats.t.fit(vpip_all2.VPIP.values)
fig = plt.figure(figsize = (20,16))
plt.suptitle('9 Randomly Selected t-distributions vs. Empirical Data \n Looking at Distribution of VP\$IP', 
             fontsize = 14, fontweight = 'bold')
for i in range(1,10):
    compare_random_t(vpip_all2,i, t_params)
plt.show()
```

```python
def buckets(x):
    if x < 51:
        return '0-50'
    elif x < 251:
        return '51-250'
    elif x < 1001:
        return '251-1000'
    elif x < 5001:
        return '1001-5000'
    elif x < 10001:
        return '5001-10,000'
    elif x < 25001:
        return '10,001-25,000'
    elif x < 100001:
        return '25,001-100,000'
    else:
        return '100k+'

vpip_all['HAND_GROUP'] = vpip_all.cnt_hands.apply(buckets)

vpip_all.head()
```

```python
hand_groups = vpip_all.groupby('HAND_GROUP').mean()

hand_groups['MIN'] = (0, 10001, 1001, 100000, 25001, 251, 5001, 51)
hand_groups = hand_groups.sort('MIN')
del hand_groups['MIN']

hand_groups
```

There are two obvious trends in this chart which speak to regulars being the ones with higher hand counts:

* the VP$IP trends down until 5000 hands, at which point it plateaus.
* the gap between VP\$IP and PFR is a well-known indicator of whether someone is a recreational player. Better players tend to be "tight and aggressive", meaning lower VP\$IP and higher PFR. Notice the gap closing as the hands played increase. 

There is also an obvious trend in terms of winrate. Before going down this path, it is worth acknowledging that there is selection bias here. A player who loses his money quickly is less likely to keep playing than someone who wins at first. However, regulars generally don't take this approach, so it's much more likely that someone who loses money then quits after few hands is a recreational player.

```python
def gmm_df(vpip_all, n, cov):
    # shuffle the dataframe
    vpip_all['RAND'] = 0
    vpip_all['RAND'] = vpip_all['RAND'].apply(lambda x: np.random.normal(0,1))
    vpip_all = vpip_all.sort('RAND')

    x1 = vpip_all.VPIP.values
    x2 = vpip_all.PFR.values

    cutoff = int(np.ceil(len(x1)*0.75))

    vpip_all['SET'] = ''
    j = 0
    for i in vpip_all.index:
        if j < cutoff:
            vpip_all.SET.ix[i] = 'Training'
        else:
            vpip_all.SET.ix[i] = 'Test'
        j += 1

    X_train = np.c_[x1[:cutoff],x2[:cutoff]]
    X_test = np.c_[x1[cutoff:],x2[cutoff:]]


    clf = mixture.GMM(n_components=n, covariance_type=cov)
    clf.fit(X_train)

    train_predict = clf.predict(X_train)
    test_predict = clf.predict(X_test)

    X_train = np.c_[X_train,train_predict]
    X_test = np.c_[X_test,test_predict]


    all_predictions = np.r_[X_train,X_test]

    vpip_all['Class'] = ''

    labels = all_predictions[:,2]
    j = 0
    for i in vpip_all.index:
        vpip_all.Class.ix[i] = labels[j]
        j += 1

    del vpip_all['RAND']
    return vpip_all

    
def gmm_plot(vpip_all, n, cov, set_type, sub, titlefont, slim = True):
    
    ax = plt.subplot(sub[0],sub[1],sub[2])

    train = vpip_all[vpip_all.SET == 'Training']
    test = vpip_all[vpip_all.SET == 'Test']

    colors = brewer2mpl.get_map('Set1', 'qualitative', 9).mpl_colors

    winrates = {}
    for i in range(n):
        if set_type == 'Training':
            class_type = train[train.Class == i]
        elif set_type == 'Test':
            class_type = test[test.Class == i]
        else:
            class_type = vpip_all[vpip_all.Class == i]
            
        class_type['TOT_EV'] = class_type.EV_250*class_type.cnt_hands/250
        win = class_type.TOT_EV.sum() / class_type.cnt_hands.sum()*250
        winrates[i] = win
        
        scatter(class_type.VPIP, class_type.PFR, color = colors[i], alpha = 0.5, 
                label = 'CL ' + str(i) + ': ' + str('%.2f' % win) + ' BB/250')

      
    plt.xlabel('VP$IP')
    plt.ylabel('PFR')

    for label in ax.get_xticklines():
        label.set_visible(False)
    for label in ax.get_yticklines():
        label.set_visible(False)
    if sub[2] != 1:
        for label in ax.get_xticklabels():
            label.set_visible(False)
        for label in ax.get_yticklabels():
            label.set_visible(False)


#     xlim([test.VPIP.min(),test.VPIP.max()])
#     ylim([test.PFR.min(),test.PFR.max()])
    xlim(10,70)
    if slim == True:
        ylim([0,80])
    else:
        ylim([0,50])
    title('Set Type: ' + set_type + ' | # of Clusters: ' + str(n) + ' | Covariance: ' + cov,
          fontsize = titlefont)

    ax.legend(loc = 1)
    return winrates
    

n = 3
cov = 'full'
set_type = 'Full'
vpip_gmm = gmm_df(vpip_all,n, cov)
fig = plt.figure(figsize = (15,10))
gmm_plot(vpip_gmm, n, cov, set_type, (1,1,1), 14, False)
plt.suptitle('Gaussian Mixture Model for Features VP\$IP and PFR', 
             fontsize = 18, color = '#262626', fontweight = 'bold')
plt.show()
```

Another way to visualize this is by looking at the regression curves for the data. The next plot shows the 2nd degree polynomial regression curves for different hand cutoffs, with error bands of 2 times the standard error of the slope.

```python
def plot_regression(df, hands, plot_types, degree, n):
    scatters = df[df.cnt_hands > hands]
    colors = brewer2mpl.get_map('BrBG', 'diverging', 8).mpl_colors

    if degree == 2:
        res = smf.ols('EV_250 ~ 1 + VPIP + I(VPIP**2.0)', scatters).fit()
    elif degree == 1:
        res = smf.ols('EV_250 ~ 1 + VPIP', scatters).fit()
    else:
        print 'only handles 1 or 2 degrees.'
        
    ax = plt.subplot(3,2,n)

    xmin = 10
    xmax = 65

    x = pd.DataFrame({'VPIP': scatters.sort('VPIP').VPIP.values})
    xx = x.VPIP.values

    if 'Data' in plot_types:
        scatter(scatters.VPIP,scatters.EV_250, color = colors[5],
                alpha = 0.5, label = 'Empirical Data')

        
    if 'Line' in plot_types:
        if 'Band' in plot_types:
            special.errorfill(xx, res.predict(x), res.bse[0]*2, label='Regression',
                      label_fill='Regression Error', color = '#262626', alpha_fill = 0.1)
        else:
            plot(xx, res.predict(x), color = colors[7], linewidth = 3,
                 label = 'Regression Curve')

    xlabel('VP$IP', fontsize = 10)
    ylabel('BB/250', fontsize = 10)
    ax.set_title('Minimum ' + str(hands) + ' hands | $R^2$ = '
                 + str('% .3f ' % res.rsquared)
                 , fontsize = 10, fontweight = 'bold')


    if 'Fake Data' in plot_types:
        int_error =  res.bse[0]
        slope_error = res.bse[1]
        b0_ols = res.params[0]
        b1_ols = res.params[1]
        if degree == 2:
            b2_ols = res.params[2]
            poly_error = res.bse[2]


        yy = np.zeros(len(xx))
        for i in range(len(xx)):
            b0 = b0_ols + np.random.normal(0,int_error)
            b1 = b1_ols + np.random.normal(0,slope_error)
            yy[i] = b0+xx[i]*b1
            if degree == 2:
                b2 = b2_ols + np.random.normal(0,poly_error)
                yy[i] += xx[i]**2*b2

        plt.scatter(xx,yy, color = colors[7], label = 'Randomly Generated Data')

    for label in ax.get_xticklines():
        label.set_visible(False)
    for label in ax.get_yticklines():
        label.set_visible(False)
    if n != 1:
#         for label in ax.get_xticklabels():
#             label.set_visible(False)
        for label in ax.get_yticklabels():
            label.set_visible(False)
        
    xlim([15,65])
    ylim([-100,100])
    ax.legend(loc = 'best')

fig = plt.figure(figsize = (20,15))
plt.suptitle('VP$IP vs. Winrate For Different Minimum Hand Cutoffs - Constant Scale',
             fontsize = 16, color = '#262626', fontweight = 'bold')
plot_types = ('Line','Data','Band') #types:('Line','Data','Fake Data','Band')
degree = 2
n = 1
for i in (350, 1000, 4000, 10000, 25000, 100000):
    plot_regression(vpip_all2, i, plot_types, degree, n)
    n += 1
```

We saw in class that a good way to plot regression is by simulating random points. Using the function built in the above cell, I generated the random data as follows:

* For all players' VP\$IP:
* Take $\beta_1, \beta_2, \beta_3$ weights from the regression.
* Add uncertainty to the weights using the distribution $err_{i,n} \sim \mathcal{N}(0,se)$ where the SD is the standard error for weight $n$.
* Generate a corresponding y for each x: $(\beta_0)(1+err_0) + (\beta_1)(1+err_1)x + (\beta_2)(1+err_2)x^2$

```python
fig = plt.figure(figsize = (20,15))
plt.suptitle('VP$IP vs. Winrate For Different Minimum Hand Cutoffs - Constant Scale '
             '\n With Randomly Generated Data',
             fontsize = 16, color = '#262626', fontweight = 'bold')
plot_types = ('Data','Fake Data') #types:('Line','Data','Fake Data','Band')
degree = 2
n = 1
for i in (350, 1000, 4000, 10000, 25000, 100000):
    plot_regression(vpip_all2, i, plot_types, degree, n)
    n += 1
```

```python
def plot_regression_agg_pfr(df, hands, plot_types, degree, n):
    scatters = vpip_all2[vpip_all2.cnt_hands > hands]
    scatters['PFR_BUCKET'] = scatters.PFR.apply(lambda x: np.round(x))
    scatters['EV'] = scatters.EV_250*scatters.cnt_hands / 250
    PFR_buckets = scatters.groupby('PFR_BUCKET').sum()
    PFR_buckets['WINRATE'] = PFR_buckets.EV / PFR_buckets.cnt_hands * 250
    PFR_buckets['PFR'] = PFR_buckets.index
    
    scatters = PFR_buckets[['PFR','WINRATE']]
    scatters.PFR = scatters.PFR.apply(lambda x: float(x))
    
    colors = brewer2mpl.get_map('YlGnBu', 'sequential', 8).mpl_colors

    if degree == 2:
        res = smf.ols('WINRATE ~ 1 + PFR + I(PFR**2.0)', scatters).fit()
    elif degree == 1:
        res = smf.ols('WINRATE ~ 1 + PFR', scatters).fit()
    else:
        print 'only handles 1 or 2 degrees.'
        
    ax = plt.subplot(3,2,n)

    xmin = 10
    xmax = 65

#     x = pd.DataFrame({'PFR': np.linspace(np.floor(xmin), np.floor(xmax), len(scatters))})
    x = pd.DataFrame({'PFR': scatters.sort('PFR').PFR.values})
    xx = x.PFR.values


    if 'Data' in plot_types:
        scatter(scatters.PFR,scatters.WINRATE, color = colors[3],
                alpha = 0.5, label = 'Empirical Data')

        
    if 'Line' in plot_types:
        if 'Band' in plot_types:
            special.errorfill(xx, res.predict(x), res.bse[0]*2, label='Regression',
                      label_fill='Regression Error', color = '#0c2c84', alpha_fill = 0.1)
        else:
            plot(xx, res.predict(x), color = colors[7], linewidth = 3,
                 label = 'Regression Curve')



    xlabel('PFR', fontsize = 10)
    ylabel('BB/250', fontsize = 10)
    ax.set_title('Minimum ' + str(hands) + ' hands | $R^2$ = '
                 + str('% .3f ' % res.rsquared)
                 , fontsize = 10, fontweight = 'bold')


    if 'Fake Data' in plot_types:
        int_error =  res.bse[0]
        slope_error = res.bse[1]
        b0_ols = res.params[0]
        b1_ols = res.params[1]
        if degree == 2:
            b2_ols = res.params[2]
            poly_error = res.bse[2]

        xx = vpip_all2[vpip_all2.cnt_hands > hands].PFR.values
        yy = np.zeros(len(xx))
        for i in range(len(xx)):
            b0 = b0_ols + np.random.normal(0,int_error)
            b1 = b1_ols + np.random.normal(0,slope_error)
            yy[i] = b0+xx[i]*b1
            if degree == 2:
                b2 = b2_ols + np.random.normal(0,poly_error)
                yy[i] += xx[i]**2*b2

        plt.scatter(xx,yy, color = colors[7], label = 'Randomly Generated Data')

    for label in ax.get_xticklines():
        label.set_visible(False)
    for label in ax.get_yticklines():
        label.set_visible(False)
    if n != 1:
#         for label in ax.get_xticklabels():
#             label.set_visible(False)
        for label in ax.get_yticklabels():
            label.set_visible(False)
        

    xlim([0,40])
    ylim([-100,100])
    ax.legend(loc = 'best')

fig = plt.figure(figsize = (20,15))
plt.suptitle('PFR vs. Average Winrate For Different Minimum Hand Cutoffs',
             fontsize = 16, color = '#262626', fontweight = 'bold')
plot_types = ('Data','Line','Band') #types:('Line','Data','Fake Data','Band')
degree = 1
n = 1
for i in (350, 1000, 4000, 10000, 25000, 100000):
    plot_regression_agg_pfr(vpip_all2, i, plot_types, degree, n)
    n += 1
```

```python
from sklearn.cluster import KMeans

def cluster_plot(scatters, clusters, min):
    np.random.seed(5)
    
    centers = [[1, 1], [-1, -1], [1, -1]]
    
    X = scatters[['VPIP','PFR']]
    
    est = KMeans(n_clusters=clusters)
    est.fit(X)
    labels = est.labels_
    
    labels = pd.Series(labels, index = X.index, name = 'LABEL')
    
    X = X.join(labels, how = 'left')
    X = X.join(scatters.EV_TOTAL, how = 'left')
    X = X.join(scatters.cnt_hands, how = 'left')

    xmin = X.VPIP.min()
    xmax = X.VPIP.max()
    ymin = X.PFR.min()
    ymax = X.PFR.max()

    ax = plt.subplot(4,2,clusters-min+1)
    fig.subplots_adjust(hspace=0.4)

    colors = brewer2mpl.get_map('Set1', 'qualitative', 9).mpl_colors
    
    for i in range(clusters):
        scatter(X[X.LABEL == i].VPIP, X[X.LABEL == i].PFR, color = colors[i], alpha = 0.5, 
                label = 'EV/250: ' + str('%.2f' % (X[X.LABEL == i].EV_TOTAL.sum() / X[X.LABEL == i].cnt_hands.sum()*250)))
    
    xlabel('VP$IP')
    ylabel('PFR')
    ax.set_title(str(clusters) + ' Clusters', fontsize = 10, fontweight = 'bold')
    ax.legend(loc = 'best')
    
    xlim([xmin,xmax])
    ylim([ymin,ymax])
    
    for label in ax.get_xticklines():
        label.set_visible(False)
    for label in ax.get_yticklines():
        label.set_visible(False)
    

fig = plt.figure(figsize = (20,16))
min = 2
plt.suptitle('VP$IP vs. PFR For Varying Cluster Sizes \n Legend Shows Weighted Average Winrate', 
             fontsize = 16, color = '#262626', fontweight = 'bold')
for i in range(min,9):
    cluster_plot(scatters, i, min)
plt.show()
```
